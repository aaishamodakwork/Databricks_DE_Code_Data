```python
data1 = [(1,'Nilesh','55000'),(2,'Nita','90000')]
columns = ['id','name','salary']
df = spark.createDataFrame(data = data1, schema = columns)
df.show()
```

```python
df.printSchema()
```

```python
from pyspark.sql.functions import col

df1 = df.withColumn(colName='salary1', col=col('salary').cast('integer'))
df1.show()
```

```python
from pyspark.sql.functions import col

df1 = df.withColumn(colName='New_col_salary', col=col('salary').cast('Integer'))
df1.show()
```

```python
df1.printSchema()
```

```python
from pyspark.sql.functions import col

df1 = df.withColumn('salary1', col('salary').cast('Integer')) \
        .withColumn('salary2', col('salary').cast('float'))

df1.show()
```

```python
df1.printSchema()
```
