```python
control_path = "s3://bizf1project123/Control/Aaisha_increment/batch_control"
```

```python
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    TimestampType
)

from pyspark.sql import functions as F
from delta.tables import DeltaTable
```

```python
control_schema_definition = StructType([
    StructField("batch_id", StringType(), False),
    StructField("status", StringType(), True),
    StructField("created_timestamp", TimestampType(), True),
    StructField("updated_timestamp", TimestampType(), True)
])

if not DeltaTable.isDeltaTable(spark, control_path):
    empty_control_df = spark.createDataFrame(
        [],
        control_schema_definition
    )

    (
        empty_control_df.write
        .format("delta")
        .mode("overwrite")
        .save(control_path)
    )

    print("Created batch control Delta table.")
else:
    print("Batch control Delta table already exists.")
```

```python
batch_control_df = spark.read.format("delta").load(control_path)

display(batch_control_df)
```
