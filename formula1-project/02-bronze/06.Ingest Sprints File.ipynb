```python
# SOURCE AND TARGET PATHS

source_file = f"{landing_folder_path}/sprints/*.json"
target_path = f"{s3_root_path}/Bronze/Aaisha/sprints"

print("Source File :", source_file)
print("Target Path :", target_path)
```

```python
from pyspark.sql.types import StructType, StructField, StringType, LongType, DoubleType

sprints_schema = StructType([
    StructField("constructorId", StringType()),
    StructField("date", StringType()),
    StructField("driverId", StringType()),
    StructField("grid", LongType()),
    StructField("laps", LongType()),
    StructField("number", LongType()),
    StructField("points", DoubleType()),
    StructField("position", LongType()),
    StructField("positionText", StringType()),
    StructField("raceName", StringType()),
    StructField("round", LongType()),
    StructField("season", LongType()),
    StructField("status", StringType()),
    StructField("url", StringType())
])
```

```python
sprints_df = (
    spark.read
        .format("json")
        .option("multiLine", "true")
        .option("mode", "FAILFAST")
        .schema(sprints_schema)
        .load(source_file)
)

display(sprints_df)
```

```python
sprints_final_df = add_ingestion_metadata(sprints_df)
```

```python
(
    sprints_final_df
        .write
        .format("delta")
        .mode("overwrite")
        .save(target_path)
)
```
